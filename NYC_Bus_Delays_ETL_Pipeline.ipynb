```python
df = df.with_columns(
    pl.col("How_Long_Delayed_Clean")
    .str.extract(r"^(\d+)-(\d+)MIN$", 1)
    .cast(pl.Int64, strict=False)
    .alias("Delay_Min_Minutes")
)

df.select([
    "How_Long_Delayed_Clean",
    "Delay_Min_Minutes"
]).filter(
    pl.col("Delay_Min_Minutes").is_not_null()
).head(10)
```

```python
df = df.with_columns(
    pl.col("How_Long_Delayed_Clean")
    .str.extract(r"^(\d+)-(\d+)MIN$", 2)
    .cast(pl.Int64, strict=False)
    .alias("Delay_Max_Minutes")
)

df.select([
    "How_Long_Delayed_Clean",
    "Delay_Min_Minutes",
    "Delay_Max_Minutes"
]).filter(
    pl.col("Delay_Max_Minutes").is_not_null()
).head(10)
```

```python
df = df.with_columns(
    pl.when(
        pl.col("How_Long_Delayed_Clean")
        .str.contains(r"^\d+MIN$")
    )
    .then(
        pl.col("How_Long_Delayed_Clean")
        .str.extract(r"^(\d+)MIN$", 1)
        .cast(pl.Int64)
    )
    .otherwise(pl.col("Delay_Min_Minutes"))
    .alias("Delay_Min_Minutes"),

    pl.when(
        pl.col("How_Long_Delayed_Clean")
        .str.contains(r"^\d+MIN$")
    )
    .then(
        pl.col("How_Long_Delayed_Clean")
        .str.extract(r"^(\d+)MIN$", 1)
        .cast(pl.Int64)
    )
    .otherwise(pl.col("Delay_Max_Minutes"))
    .alias("Delay_Max_Minutes")
)
```

```python
df = df.with_columns(
    pl.when(
        pl.col("How_Long_Delayed_Clean")
        .str.contains(r"^\d+HR$")
    )
    .then(
        pl.col("How_Long_Delayed_Clean")
        .str.extract(r"^(\d+)HR$", 1)
        .cast(pl.Int64, strict=False) * 60
    )
    .otherwise(pl.col("Delay_Min_Minutes"))
    .alias("Delay_Min_Minutes"),

    pl.when(
        pl.col("How_Long_Delayed_Clean")
        .str.contains(r"^\d+HR$")
    )
    .then(
        pl.col("How_Long_Delayed_Clean")
        .str.extract(r"^(\d+)HR$", 1)
        .cast(pl.Int64, strict=False) * 60
    )
    .otherwise(pl.col("Delay_Max_Minutes"))
    .alias("Delay_Max_Minutes")
)
```

```python
string_cols = [
    col for col, dtype in df.schema.items()
    if dtype == pl.String
]

df = df.with_columns(
    [
        pl.col(col)
        .str.strip_chars()
        .alias(col)
        for col in string_cols
    ]
)

print("Cleaned string columns:", len(string_cols))
```

```python
df = df.with_columns([
    pl.col("Occurred_On").dt.year().alias("Incident_Year"),
    pl.col("Occurred_On").dt.month().alias("Incident_Month"),
    pl.col("Occurred_On").dt.day().alias("Incident_Day"),
    pl.col("Occurred_On").dt.hour().alias("Incident_Hour"),
    pl.col("Occurred_On").dt.weekday().alias("Incident_Weekday")
])
```

```python
df.select([
    "Occurred_On",
    "Incident_Year",
    "Incident_Month",
    "Incident_Day",
    "Incident_Hour",
    "Incident_Weekday"
]).head()
```

```python
df.select([
    pl.len().alias("total_rows"),
    pl.col("Busbreakdown_ID").n_unique().alias("unique_busbreakdown_ids"),
    pl.col("Occurred_On").is_null().sum().alias("null_occurred_on"),
    pl.col("Number_Of_Students_On_The_Bus").is_null().sum().alias("null_student_count"),
    pl.col("Delay_Avg_Minutes").is_null().sum().alias("null_delay_minutes")
])
```

```python
df = df.with_columns(
    pl.when(
        pl.col("Occurred_On").is_null() |
        pl.col("Created_On").is_null()
    )
    .then(pl.lit("MISSING_DATE"))

    .when(
        pl.col("Occurred_On") <= pl.col("Created_On")
    )
    .then(pl.lit("VALID"))

    .when(
        pl.col("Occurred_On").dt.date() ==
        pl.col("Created_On").dt.date()
    )
    .then(pl.lit("SAME_DAY_ORDER_ISSUE"))

    .otherwise(pl.lit("DATE_SEQUENCE_ISSUE"))
    .alias("Date_Quality_Flag")
)

df.group_by("Date_Quality_Flag").len().sort("len", descending=True)
```

```python
# Overwrite the Parquet file with the final cleaned DataFrame
df.write_parquet(
    "bus_delays_clean.parquet",
    compression="snappy"
)

print("Final Parquet file updated")
```

```python
from google.cloud import bigquery

project_id = "electric-nomad-388815"

client = bigquery.Client(project=project_id)

print("Connected to BigQuery project:", project_id)
```

```python
dataset_id = f"{project_id}.transportation_data"

dataset = bigquery.Dataset(dataset_id)
dataset.location = "US"

dataset = client.create_dataset(dataset, exists_ok=True)

print("Dataset ready:", dataset_id)
```

```python
table_id = f"{project_id}.transportation_data.bus_delays"

job_config = bigquery.LoadJobConfig(
    source_format=bigquery.SourceFormat.PARQUET,
    write_disposition=bigquery.WriteDisposition.WRITE_TRUNCATE
)

with open("bus_delays_clean.parquet", "rb") as source_file:
    load_job = client.load_table_from_file(
        source_file,
        table_id,
        job_config=job_config
    )

load_job.result()

print("Parquet loaded successfully into BigQuery")
```

```python
query = f"""
SELECT COUNT(*) AS total_rows
from {table_id}
"""
result = client.query(query).to_dataframe()
result
```

```python
query = f"""
SELECT
    Reason,
    COUNT(*) AS total_incidents
FROM `{table_id}`
WHERE Reason IS NOT NULL
GROUP BY Reason
ORDER BY total_incidents DESC
LIMIT 10
"""

result = client.query(query).to_dataframe()

result
```

```python
query = f"""
SELECT
    Boro,
    COUNT(*) AS total_incidents
FROM `{table_id}`
WHERE Boro IS NOT NULL
GROUP BY Boro
ORDER BY total_incidents DESC
"""

borough_result = client.query(query).to_dataframe()

borough_result
```

```python
query = f"""
SELECT
    Incident_Hour,
    COUNT(*) AS total_incidents
FROM `{table_id}`
WHERE Incident_Hour IS NOT NULL
GROUP BY Incident_Hour
ORDER BY Incident_Hour
"""

hour_result = client.query(query).to_dataframe()

hour_result
```

```python
query = f"""
SELECT
    Reason,
    ROUND(AVG(Delay_Avg_Minutes), 2) AS avg_delay_minutes,
    COUNT(*) AS total_incidents
FROM `{table_id}`
WHERE
    Reason IS NOT NULL
    AND Delay_Avg_Minutes IS NOT NULL
GROUP BY Reason
HAVING COUNT(*) >= 100
ORDER BY avg_delay_minutes DESC
LIMIT 10
"""

delay_reason_result = client.query(query).to_dataframe()

delay_reason_result
```

```python
query = f"""
SELECT
    Incident_Year,
    COUNT(*) AS total_incidents
FROM `{table_id}`
WHERE Incident_Year IS NOT NULL
GROUP BY Incident_Year
ORDER BY Incident_Year
"""

year_result = client.query(query).to_dataframe()

year_result
```

```python
query = f"""
SELECT
    School_Year,
    Occurred_On,
    Incident_Year,
    Busbreakdown_ID,
    Reason
FROM `{table_id}`
WHERE Incident_Year < 2015
   OR Incident_Year > 2026
ORDER BY Incident_Year
LIMIT 30
"""

date_quality_check = client.query(query).to_dataframe()

date_quality_check
```

```python
query = f"""
SELECT
    School_Year,
    Busbreakdown_ID,
    Occurred_On,
    Created_On,
    Last_Updated_On,
    Reason,
    Boro
FROM `{table_id}`
WHERE Incident_Year = 2027
"""

future_date_check = client.query(query).to_dataframe()

future_date_check
```

```python
query = f"""
SELECT
    COUNT(*) AS invalid_date_sequence
FROM `{table_id}`
WHERE Occurred_On > Created_On
"""

date_issue_count = client.query(query).to_dataframe()

date_issue_count
```

```python
query = f"""
SELECT
    CASE
        WHEN DATE_DIFF(DATE(Occurred_On), DATE(Created_On), DAY) = 0
            THEN 'Same day'
        WHEN DATE_DIFF(DATE(Occurred_On), DATE(Created_On), DAY) BETWEEN 1 AND 7
            THEN '1-7 days'
        WHEN DATE_DIFF(DATE(Occurred_On), DATE(Created_On), DAY) BETWEEN 8 AND 365
            THEN '8-365 days'
        ELSE 'More than 1 year'
    END AS date_gap,
    COUNT(*) AS records
FROM `{table_id}`
WHERE Occurred_On > Created_On
GROUP BY date_gap
ORDER BY records DESC
"""

gap_result = client.query(query).to_dataframe()

gap_result
```

```python
query = f"""
SELECT
    Date_Quality_Flag,
    COUNT(*) AS records
FROM `{table_id}`
GROUP BY Date_Quality_Flag
ORDER BY records DESC
"""

quality_result = client.query(query).to_dataframe()

quality_result
```

## Project Summary

Built an end-to-end ETL pipeline using NYC Bus Breakdown and Delays open data.

### Pipeline
CSV → Polars → Parquet → Google BigQuery → SQL Analytics

### Key Work Completed
- Processed 1,297,681 records using Polars.
- Cleaned inconsistent text and numeric fields.
- Converted timestamp columns to proper datetime types.
- Standardized delay-duration values and created numeric delay metrics.
- Created derived date features for analysis.
- Added data-quality validation for date inconsistencies.
- Converted the cleaned dataset to Parquet using Snappy compression.
- Loaded the final dataset into Google BigQuery.
- Validated 1,297,681 rows and 31 columns in BigQuery.
- Performed SQL analysis on delay reasons, boroughs, incident hours, delay duration, and yearly trends.

### Data Quality Results
- VALID: 1,268,884
- SAME_DAY_ORDER_ISSUE: 28,731
- DATE_SEQUENCE_ISSUE: 65
- MISSING_DATE: 1
